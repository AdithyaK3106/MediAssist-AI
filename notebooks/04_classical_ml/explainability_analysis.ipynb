# Explainability Analysis

This notebook explores feature importance for classical ML models, primarily focusing on TF-IDF word weights for Logistic Regression and Trees.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import os
import numpy as np
```

## Feature Importance (Logistic Regression)

```python
# Load model and vectorizer
model_path = '../../models/classical/logistic_regression/classifier.pkl'
vectorizer_path = '../../models/classical/logistic_regression/vectorizer.pkl'

if os.path.exists(model_path) and os.path.exists(vectorizer_path):
    model = joblib.load(model_path)
    vectorizer = joblib.load(vectorizer_path)
    
    # Get feature names
    feature_names = vectorizer.get_feature_names_out()
    
    # Average coefficients across all classes for simplicity
    avg_importance = np.mean(np.abs(model.coef_), axis=0)
    
    # Create DataFrame
    importance_df = pd.DataFrame({'Feature': feature_names, 'Importance': avg_importance})
    importance_df = importance_df.sort_values(by='Importance', ascending=False).head(20)
    
    plt.figure(figsize=(10, 6))
    sns.barplot(data=importance_df, x='Importance', y='Feature')
    plt.title('Top 20 Features (TF-IDF Words) - Logistic Regression')
    plt.tight_layout()
    plt.show()
else:
    print('Model files not found. Please train the logistic regression model first.')
```
