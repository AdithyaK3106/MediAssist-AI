# Classical ML Model Comparison

Comparing benchmark performance across Logistic Regression, Naive Bayes, Decision Trees, Random Forests, and XGBoost.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

## Load Leaderboard

```python
leaderboard_file = '../../reports/model_benchmarks/leaderboard.csv'
if os.path.exists(leaderboard_file):
    leaderboard = pd.read_csv(leaderboard_file)
    display(leaderboard)
    
    plt.figure(figsize=(10, 5))
    sns.barplot(data=leaderboard.sort_values('f1_score', ascending=False), x='model_name', y='f1_score', palette='viridis')
    plt.title('F1 Score Comparison Across Models')
    plt.ylabel('F1 Score')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print('Leaderboard not found.')
```
