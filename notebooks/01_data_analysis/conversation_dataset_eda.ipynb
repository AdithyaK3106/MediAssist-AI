# Conversation Dataset EDA
NLP-centric analysis of patient-doctor medical interactions.

## 1. Environment and Data Loading

```python
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
%matplotlib inline

sns.set(style='whitegrid')
```

```python
RAW_PATH = '../../data/raw/conversations/HealthCareMagic-100k.json'
with open(RAW_PATH, 'r') as f:
    data = json.load(f)
df = pd.DataFrame(data)
print(f'Total Conversations: {len(df)}')
df.head()
```

## 2. Text Length Distribution

### Pre-generated Length Distribution
![Length Distribution](../../reports/figures/conv_length_dist.png)

```python
df['input_len'] = df['input'].apply(lambda x: len(str(x).split()))
df['output_len'] = df['output'].apply(lambda x: len(str(x).split()))

plt.figure(figsize=(12, 6))
sns.histplot(df['input_len'], bins=100, color='blue', label='Patient Input', kde=True)
sns.histplot(df['output_len'], bins=100, color='green', label='Doctor Output', kde=True)
plt.xlim(0, 500)
plt.title('Token Count Distribution')
plt.legend()
plt.show()
```

## 3. Linguistic Visualization: Word Clouds

### Pre-generated Word Cloud
![Input Wordcloud](../../reports/figures/conv_wordcloud_input.png)

```python
sample_text = ' '.join(df['input'].astype(str).sample(5000))
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(sample_text)

plt.figure(figsize=(15, 8))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Patient Query Word Cloud')
plt.show()
```

## 4. Vocabulary Analysis

### Pre-generated Top Words
![Top Words](../../reports/figures/conv_top_words.png)

```python
from collections import Counter
import re

def get_tokens(text):
    return re.findall(r'\w+', str(text).lower())

all_tokens = []
for t in df['input'].sample(10000): all_tokens.extend(get_tokens(t))

# Filter common stopwords
sw = {'the', 'and', 'i', 'to', 'a', 'is', 'have', 'of', 'in', 'it', 'my', 'for'}
filtered = [w for w in all_tokens if w not in sw]

counts = Counter(filtered)
top_20 = pd.DataFrame(counts.most_common(20), columns=['Word', 'Freq'])
sns.barplot(data=top_20, x='Freq', y='Word', palette='magma')
plt.title('Top 20 Frequent Medical Terms (Filtered)')
plt.show()
```

## 5. NLP Model Readiness (BERT/DistilBERT)
- **Realism**: High variety in expression supports robust NLU.
- **Symmetry**: Balanced input/output lengths suggest good sequence-to-sequence potential.
- **Noise**: Typos and shorthand exist, requiring robust tokenization (WordPiece).
