# Hospital Dataset EDA
Healthcare infrastructure and availability analysis.

## 1. Data Loading
Reading the national hospital directory and local rural health datasets.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
HOSP_PATH = '../../data/raw/hospitals/hospital_directory.csv'
RURAL_PATH = '../../data/raw/rural_health/Health_Care_Facilities_Tiruppur_0.csv'

try:
    df_hosp = pd.read_csv(HOSP_PATH, encoding='utf-8')
except:
    df_hosp = pd.read_csv(HOSP_PATH, encoding='cp1252')

df_rural = pd.read_csv(RURAL_PATH)
print('Data loaded successfully.')
```

## 2. Geographic Distribution

### Pre-generated State Distribution
![State Distribution](../../reports/figures/hosp_state_dist.png)

```python
df_hosp['State'].value_counts().plot(kind='bar', figsize=(15, 7), color='teal')
plt.title('Hospital Density by State')
plt.ylabel('Count')
plt.show()
```

## 3. Hospital Categorization

### Pre-generated Category Distribution
![Category Distribution](../../reports/figures/hosp_category_pie.png)

```python
df_hosp['Hospital_Category'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(10, 10))
plt.title('Facility Ownership Distribution')
plt.ylabel('')
plt.show()
```

## 4. Rural Facility Geo-Analysis

### Pre-generated Geo Distribution
![Geo Distribution](../../reports/figures/rural_geo_dist.png)

```python
if 'LATITUDE' in df_rural.columns:
    plt.figure(figsize=(10, 8))
    sns.scatterplot(data=df_rural, x='LONGITUDE', y='LATITUDE', hue='Govt and Private', alpha=0.7)
    plt.title('Tiruppur Rural Health Facility Distribution')
    plt.show()
```

- **Geo-Precision**: Local rural data is much richer in coordinate data than the national set.
